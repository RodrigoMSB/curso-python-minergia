## Paso 5 de 8 · ¿Cómo guardo varios valores juntos?

**Qué vamos a responder.** Cómo se guardan varios valores juntos.

**Qué hacer.** Corre la celda. Una lista guarda varios precios bajo una sola etiqueta.

```python
# Paso 5
# Una lista va entre corchetes, con los valores separados por comas.
precios = [1500, 990, 2300]
print(precios)
print("Cantidad de productos", len(precios))
print("Suma de los precios", sum(precios))
```

## Paso 6 de 8 · ¿Cómo decide Python?

**Qué vamos a responder.** Cómo elige Python entre dos caminos.

**Qué hacer.** Mueve el deslizador del stock y corre la celda.

```python
# Paso 6
# if pregunta algo. Si la respuesta es sí, hace lo de abajo.
# Si no, hace lo del else.
stock_pan = 0 #@param {type:"slider", min:0, max:10, step:1}

if stock_pan > 0:
    print("Hay pan. Compro pan.")
else:
    print("No hay pan. Compro galletas.")
```

## Paso 7 de 8 · ¿Cómo repite Python una tarea?

**Qué vamos a responder.** Cómo repite Python una tarea para cada valor de una lista.

**Qué hacer.** Corre la celda. Pasamos los tres productos por la caja, uno por uno.

```python
# Paso 7
# for repite lo de abajo una vez por cada valor de la lista.
for precio in precios:
    print("Pasa por la caja un producto de", precio)
```
