## Paso 5 de 8 · ¿Una central es chica, mediana o grande?

**Qué vamos a responder.** Cómo decide Python en qué categoría cae una central según su potencia.

**Qué hacer.** Mueve el deslizador de potencia y vuelve a correr la celda.

```python
# Paso 5
# Mueve el deslizador y vuelve a correr la celda para ver cómo cambia la categoría.
potencia = 230 #@param {type:"slider", min:0, max:500, step:10}

# Python revisa las condiciones en orden y ejecuta solo la primera que se cumple.
# Las líneas con sangría son las que dependen de cada condición.
if potencia < 100:
    categoria = "chica"          # menos de 100 MW
elif potencia < 300:
    categoria = "mediana"        # desde 100 y hasta menos de 300 MW
else:
    categoria = "grande"         # 300 MW o más
print("Una central de", potencia, "MW es", categoria)
```

## Paso 6 de 8 · Tu turno, mueve los cortes

**Qué vamos a responder.** Cuántas centrales caen en cada categoría según dónde se pongan los cortes.

**Qué hacer.** Corre la celda con los cortes en 100 y 300 MW. Después mueve los deslizadores y vuelve a correrla.

```python
# Paso 6
# Los dos cortes que separan las categorías, en MW.
corte_chica = 100 #@param {type:"slider", min:50, max:200, step:10}
corte_grande = 300 #@param {type:"slider", min:200, max:450, step:10}

# Cada central parte como grande y baja de categoría si no alcanza el corte.
conteo = {"chica": 0, "mediana": 0, "grande": 0}
for potencia in potencias:
    categoria = "grande"
    if potencia < corte_chica:
        categoria = "chica"
    elif potencia < corte_grande:
        categoria = "mediana"
    print(potencia, "MW es", categoria)
    conteo[categoria] = conteo[categoria] + 1   # suma 1 a su categoría
print("✓ Conteo", conteo)
```

## Paso 7 de 8 · ¿Cuánta energía produce cada central en un día?

**Qué vamos a responder.** Cuántos MWh produce en un día cada una de las cinco centrales.

**Qué hacer.** Corre la celda. La tabla usa las potencias del Paso 3.

```python
# Paso 7
import pandas as pd

# Las cinco centrales en una tabla, con su potencia y su factor de planta.
tabla = pd.DataFrame({
    "central": ["Río Manso Alto", "Pampa Alta", "Cerro Negro", "Bahía Norte", "Respaldo Cordillera"],
    "potencia_mw": potencias,
    "factor_planta": [0.56, 0.25, 0.34, 0.50, 0.03],
})
# Energía de un día: potencia por 24 horas por el factor de planta.
tabla["mwh_dia"] = tabla["potencia_mw"] * 24 * tabla["factor_planta"]
tabla
```

## Paso 8 de 8 · La pregunta del lab

**Qué vamos a responder.** Cuánta energía producen en un día las centrales grandes, medianas o chicas.

**Qué hacer.** Elige una categoría en la lista y vuelve a correr la celda. Los cortes quedan fijos en 100 y 300 MW.

```python
# Paso 8
# Los cortes quedan fijos en 100 y 300 MW, los mismos del Paso 5.
categoria = "grande" #@param ["chica", "mediana", "grande"]

if categoria == "chica":
    elegidas = tabla[tabla["potencia_mw"] < 100]
elif categoria == "mediana":
    elegidas = tabla[(tabla["potencia_mw"] >= 100) & (tabla["potencia_mw"] < 300)]
else:
    elegidas = tabla[tabla["potencia_mw"] >= 300]

# La energía de un día de las centrales elegidas, sumada.
total = round(elegidas["mwh_dia"].sum(), 1)
print(f"✓ Las centrales {categoria}s producen {total} MWh en un día")
elegidas
```
