## 2. Listas

````python
# Una lista va entre corchetes, con los datos separados por comas.
# Son las potencias de las cinco centrales, en el mismo orden de la tabla.
potencias = [420, 230, 160, 375, 45]

print("Potencias", potencias)
print("Cantidad de centrales", len(potencias))   # len cuenta los elementos
print("Potencia total", sum(potencias), "MW")    # sum los suma
print("La más grande", max(potencias), "MW")     # max entrega el mayor
print("La más chica", min(potencias), "MW")      # min entrega el menor
````

````python
# Cada elemento tiene una posición. Python cuenta desde 0, no desde 1.
print("En la posición 0 está", potencias[0], "y en la posición 4 está", potencias[4])

# Un for repite las líneas con sangría una vez por cada elemento de la lista.
# numero parte en 1 y en cada vuelta le sumamos 1, así lleva la cuenta.
numero = 1
for potencia in potencias:
    print("Central", numero, "tiene", potencia, "MW")
    numero = numero + 1
````

## 4. Decisiones y repetición

````python
# Mueve el deslizador y vuelve a correr la celda para ver cómo cambia la categoría.
potencia = 230 #@param {type:"slider", min:0, max:500, step:10}

# Python revisa las condiciones en orden y ejecuta solo la primera que se cumple.
# Las líneas con sangría son las que dependen de cada condición.
if potencia < 100:
    categoria = "chica"          # menos de 100 MW
elif potencia < 300:
    categoria = "mediana"        # desde 100 y hasta menos de 300 MW
else:
    categoria = "grande"         # 300 MW o más
print("Una central de", potencia, "MW es", categoria)
````

````python
# El mismo criterio para las cinco centrales, contando cuántas caen en cada categoría.
conteo = {"chica": 0, "mediana": 0, "grande": 0}    # un contador por categoría, todos parten en 0
for central in centrales:
    if central["potencia_mw"] < 100:
        categoria = "chica"
    elif central["potencia_mw"] < 300:
        categoria = "mediana"
    else:
        categoria = "grande"
    print(central["nombre"], "es", categoria)
    conteo[categoria] = conteo[categoria] + 1       # suma 1 al contador de su categoría
print("Conteo final", conteo)
````

## 5. Funciones

````python
# def crea la función. potencia es el parámetro, el dato que recibe cada vez que la llamas.
def clasificar(potencia):
    if potencia < 100:
        return "chica"           # return entrega el resultado y termina la función
    elif potencia < 300:
        return "mediana"
    else:
        return "grande"

# Al llamarla, el valor entre paréntesis ocupa el lugar de potencia.
print("45 MW es", clasificar(45))
print("230 MW es", clasificar(230))
print("420 MW es", clasificar(420))
````

````python
# Una función puede recibir más de un parámetro, separados por coma.
# En un día de 24 horas, la central produce su potencia por el factor de planta.
def energia_dia(potencia_mw, factor_planta):
    return potencia_mw * 24 * factor_planta     # el resultado queda en MWh

# La probamos con la primera central, pasándole sus datos desde el diccionario.
primera = centrales[0]
print(primera["nombre"], "produce", energia_dia(primera["potencia_mw"], primera["factor_planta"]), "MWh en un día")
````
