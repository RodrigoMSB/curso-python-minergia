## Paso 6 de 7 · ¿Qué centrales hay en una región?

**Qué vamos a responder.** Qué centrales hay en una región y cuál es la más grande.

![Un filtro deja pasar las centrales de una región](https://raw.githubusercontent.com/RodrigoMSB/curso-python-minergia/main/modulos/bienvenida/img/bienvenida_paso6_filtro.png)

**Qué hacer.** Elige una región en la lista y vuelve a correr la celda. La celda no se corre sola al cambiar la región.

```python
# Paso 6
# Las centrales de la región elegida, de la más grande a la más chica.
region = "Biobío" #@param ["Antofagasta", "Atacama", "Coquimbo", "Valparaíso", "Metropolitana", "Biobío", "Los Lagos"]

en_la_region = centrales[centrales["region"] == region]
en_la_region = en_la_region.sort_values("potencia_mw", ascending=False)

print(f"✓ En {region} hay {len(en_la_region)} centrales")
en_la_region[["central", "tecnologia", "potencia_mw"]]
```

## Paso 7 de 7 · ¿Con qué tecnologías se genera la electricidad?

**Qué vamos a responder.** Cuánta potencia instalada tiene cada tecnología.

![La potencia de cada central sumada por tecnología](https://raw.githubusercontent.com/RodrigoMSB/curso-python-minergia/main/modulos/bienvenida/img/bienvenida_paso7_suma.png)

**Qué hacer.** Corre la celda. Esto es lo que vamos a hacer en todo el curso, hacerle preguntas a los datos y leer la respuesta.

```python
# Paso 7
# La potencia instalada de cada tecnología, de la mayor a la menor.
por_tecnologia = centrales.groupby("tecnologia")["potencia_mw"].sum()
por_tecnologia.sort_values(ascending=False).reset_index()
```
